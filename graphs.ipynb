```python
import pandas as pd

# Load the CSV file into a DataFrame
df = pd.read_csv("Housing")

# Display the first 5 rows of the DataFrame
df.head()
```

```python
# Calculate Q1, Q2, and Q3 for a specific column
q1 = df['price'].quantile(0.25)
q2 = df['price'].quantile(0.50) # Same as df['price'].median()
q3 = df['price'].quantile(0.75)

print(f"25th Percentile (Q1): {q1}")
print(f"50th Percentile (Q2): {q2}")
print(f"75th Percentile (Q3): {q3}")
```

```python
IQR = q3 - q1
print (IQR)
```

```python
lower = q1 - 1.5 * IQR
print (lower)
```

```python
# Returns True if any price is <= -35000, otherwise False
exists = (df['price'] <= -35000).any()
print("Do lower outliers exist?", exists)
```

```python
upper = q3 + 1.5 * IQR 
print (upper)
```

```python
exists = (df['price'] >= 9205000).any()
print("Do lower outliers exist?", exists)
```

```python
# Create a list of the outlier rows
outlier_list = df[df['price'] >= 9205000 ]

# Display the list in your notebook
outlier_list
```

```python
expensive_houses_count = (df['price'] > upper).sum()

print(f"Upper Threshold: {upper:,}")
print(f"Number of way expensive houses (upper outliers): {expensive_houses_count}")
```

```python
# Filter the DataFrame to show only the 15 expensive houses
expensive_houses_list = df[df['price'] > upper]

# Display the 15 houses with all their properties/columns
expensive_houses_list
```

```python
# Keep only the rows where price is within the normal range
df_cleaned = df[df['price'] <= upper]

# Check the new size of your dataset
print(f"Original number of houses: {len(df)}")
print(f"New number of houses: {len(df_cleaned)}")
print(f"Successfully removed: {len(df) - len(df_cleaned)} houses")
```
