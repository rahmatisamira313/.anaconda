```python
q1 = df["Year"].quantile(0.25)
q1
```

```python
q3 = df["Year"].quantile(0.75)
q3
```

```python
IQR = q3-q1
IQR
```

```python
Lower = q1 - 1.5 * IQR
Lower
```

```python
Upper = q3 + 1.5 * IQR
Upper
```

```python
df_clean = df[(df['Year'] >= Lower) & (df['Year'] <= Upper)]
df_clean
```

```python
df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(data=df_clean, x='Year', y='Mortality Rate')

# Add labels and title
plt.title('Scatter Plot of mortality and year')
plt.xlabel('Year')
plt.ylabel('Mortality Rate')
```

```python
# the graph above shows that form the years 1960s on the mortality rate of infants and children had been decreasing 
```
